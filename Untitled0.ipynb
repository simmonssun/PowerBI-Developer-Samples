```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
df = pd.read_csv('/content/credit_risk_dataset.csv')
print("--- Primeiras Linhas ---")
display(df.head())
```

```python
# Verificar dimensões, colunas e tipos de dados
print("\n--- Informações do Dataset ---")
df.info()
```

```python
# Estatísticas descritivas das variáveis numéricas
print("\n--- Estatísticas Descritivas ---")
display(df.describe())
```

```python
# 1. Verificar valores nulos
print("Valores ausentes por coluna:")
print(df.isnull().sum())
```

```python
# Tratamento dos nulos:
# - Preencher tempo de emprego nulo pela mediana
df['person_emp_length'] = df['person_emp_length'].fillna(df['person_emp_length'].median())
```

```python
# - Preencher taxa de juros nula pela mediana agrupada por nota do empréstimo (loan_grade)
df['loan_int_rate'] = df.groupby('loan_grade')['loan_int_rate'].transform(lambda x: x.fillna(x.median()))


# 2. Tratamento de Outliers (Inconsistências)
# Exemplo: Pessoas com mais de 100 anos de idade ou tempo de emprego irreal
df = df[df['person_age'] < 100]
df = df[df['person_emp_length'] < 60]

print("\nFormato do DataFrame após limpeza:", df.shape)
```

```python
# 1. Criar faixa etária
rotulos_idade = ['18-25', '26-35', '36-50', '50+']
limites_idade = [17, 25, 35, 50, 100]
df['faixa_etaria'] = pd.cut(df['person_age'], bins=limites_idade, labels=rotulos_idade)

# 2. Criar renda mensal estimada
df['renda_mensal'] = df['person_income'] / 12

# 3. Categorizar a variável alvo (loan_status) para melhor leitura nos gráficos
df['status_inadimplencia'] = df['loan_status'].map({0: 'Adimplente', 1: 'Inadimplente'})
```

```python
taxa_inadimplencia = df['loan_status'].mean() * 100
print(f"Taxa de Inadimplência Geral: {taxa_inadimplencia:.2f}%")

sns.countplot(data=df, x='status_inadimplencia', palette=['#2ecc71', '#e74c3c'])
plt.title('Distribuição de Adimplentes vs. Inadimplentes')
plt.ylabel('Quantidade de Clientes')
plt.xlabel('Status do Empréstimo')
plt.show()
```

```python
grade_risk = df.groupby('loan_grade')['loan_status'].mean().reset_index()
grade_risk['taxa_pct'] = grade_risk['loan_status'] * 100

sns.barplot(data=grade_risk, x='loan_grade', y='taxa_pct', palette='Reds_d')
plt.title('Taxa de Inadimplência por Nota de Risco (Grade)')
plt.xlabel('Nota do Empréstimo (A = Menor Risco, G = Maior Risco)')
plt.ylabel('Taxa de Inadimplência (%)')
plt.show()
```

```python
sns.boxplot(data=df, x='status_inadimplencia', y='loan_percent_income', palette=['#2ecc71', '#e74c3c'])
plt.title('Comprometimento da Renda por Status de Pagamento')
plt.xlabel('Status')
plt.ylabel('Proporção Empréstimo / Renda Anual')
plt.show()
```

```python
colunas_numericas = df.select_dtypes(include=[np.number])
plt.figure(figsize=(10, 8))
sns.heatmap(colunas_numericas.corr(), annot=True, fmt=".2f", cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Matriz de Correlação das Variáveis Numéricas')
plt.show()
```

```python
# Exportar base tratada para a pasta do projeto
df.to_csv('credit_risk_dataset_tratado.csv', index=False)
print("Base tratada exportada com sucesso para 'credit_risk_dataset_tratado.csv'!")
```
